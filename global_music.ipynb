```python
import pandas as pd

# Assuming the file_path is known from previous cells
file_path = "/content/drive/MyDrive/colab csv/Global_Music_Streaming_Listener_Preferences.csv"
df = pd.read_csv(file_path)

# Define categorical_features as it's used below
categorical_features = [feature for feature in df.columns if df[feature].dtype == 'O']

# Correct way to select columns from a DataFrame using a list of column names
df[categorical_features]
```

```python
#univariate Analysis
#distribution of users over age
plt.figure(figsize=(8,5))
sns.histplot(df["Age"], bins=20, kde=True, color='skyblue')
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("distribution of user age")
plt.show()
```

```python
#popular streaming platform
plt.figure(figsize=(10,4))
sns.countplot(x=df["Streaming Platform"], order=df["Streaming Platform"].value_counts().index, palette ="Set1")
plt.xlabel("Streaming Platform")
plt.ylabel("Number of users")
plt.title("Most popular streaming platform")
```

```python
top_genre =df["Top Genre"].value_counts()
plt.figure(figsize=(8,6))
plt.pie(top_genre,labels = top_genre.index, autopct='%1.1f%%')
plt.title("Favourite Genre")
plt.show()
```

```python
#favourable time of users to listen songs
plt.figure(figsize=(8,6))
sns.countplot(x=df["Listening Time (Morning/Afternoon/Night)"])
plt.xlabel("Listening Time")
plt.ylabel("Number of users")
plt.title("Favourable time of users")
```

```python
#bivariate and multivariate analysis
plt.figure(figsize=(8,6))
sns.countplot(data=df,y="Most Played Artist",hue="Subscription Type")
plt.xlabel("Number of users")
plt.ylabel("Artist")
plt.title("Artist vs Subscription type")
```

```python
#bivariate and multivariate analysis
plt.figure(figsize=(8,6))
sns.countplot(data=df,y="Most Played Artist",hue="Subscription Type")
plt.xlabel("Number of users")
plt.ylabel("Most Played Artist")
plt.title("Artist vs Subscription type")
```

```python
#distribution of music genre by age group
bins = [0,18,30,50,70,100]
labels = ['<18','18-30','31-50','51-70','71+']
df['Age_group'] = pd.cut(df['Age'], bins=bins, labels=labels, right=False)
genre_age_counts = df.groupby(["Age_group","Top Genre"], observed= False).size().unstack().fillna(0)
plt.figure(figsize=(10,6))
sns.heatmap(genre_age_counts, annot=True, fmt='d')
plt.title("Heatmap: Distribution of music genre by age group")
plt.xlabel("Age_group")
plt.ylabel("Top Genre")
plt.show()
```

```python
subscriptiontype = df.groupby("Subscription Type")["Minutes Streamed Per Day"].count()
subscriptiontype
```

create a bar chart :- between most played artist and number of songs liked

```python
artist_likes = df.groupby('Most Played Artist')['Number of Songs Liked'].sum().sort_values(ascending=False)

plt.figure(figsize=(12, 6))
sns.barplot(x=artist_likes.index, y=artist_likes.values, hue=artist_likes.index, palette='viridis', legend=False)
plt.xlabel('Most Played Artist')
plt.ylabel('Total Number of Songs Liked')
plt.title('Total Number of Songs Liked by Most Played Artist')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
